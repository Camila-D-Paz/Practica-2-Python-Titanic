**Preguntas 3:**
* ¿Cuántos **tickets** se pagaron con cada método de pago?
* ¿Cuántos tickets tienen este dato **sin registrar** (valores nulos)?

````python

contEfec = 0
contTarjet = 0
contTrans = 0
contSinn = 0

tickets = ven.drop_duplicates("id_ticket")

for metodo in tickets["metodo_pago"]:
    if metodo == "Transferencia":
        contTrans += 1
    elif metodo == "Tarjeta":
        contTarjet += 1
    elif metodo == "Efectivo":
        contEfec += 1
    else:
        contSinn += 1

print("Tickets pagados por transferencia:", contTrans)
print("Tickets pagados en efectivo:", contEfec)
print("Tickets pagados con tarjeta:", contTarjet)
print("Tickets sin método de pago registrado:", contSinn)
````

**Preguntas 4:**
* ¿Cuál es el **precio de lista mínimo, máximo y promedio** de los productos de cada categoría?

> **Nota:** Utilice el catálogo de productos para este análisis.

````python
precioE = []
precioAr = []
precioAcce = []
precioCuader = []

for categoria, precio in zip(pro["categoria"], pro["precio_lista"]):
    if categoria == "Escritura":
        precioE.append(precio)
    elif categoria == "Archivo":
        precioAr.append(precio)
    elif categoria == "Accesorios":
        precioAcce.append(precio)
    elif categoria == "Papel y cuadernos":
        precioCuader.append(precio)

promE = sum(precioE) / len(precioE)
minE = min(precioE)
maxE = max(precioE)

promAr = sum(precioAr) / len(precioAr)
minAr = min(precioAr)
maxAr = max(precioAr)

promAcce = sum(precioAcce) / len(precioAcce)
minAcce = min(precioAcce)
maxAcce = max(precioAcce)

promCuader = sum(precioCuader) / len(precioCuader)
minCuader = min(precioCuader)
maxCuader = max(precioCuader)

print("Escritura - Máximo:", round(maxE, 2), "Mínimo:", round(minE, 2), "Promedio:", round(promE, 2))
print("Archivo - Máximo:", round(maxAr, 2), "Mínimo:", round(minAr, 2), "Promedio:", round(promAr, 2))
print("Accesorios - Máximo:", round(maxAcce, 2), "Mínimo:", round(minAcce, 2), "Promedio:", round(promAcce, 2))
print("Papel y cuadernos - Máximo:", round(maxCuader, 2), "Mínimo:", round(minCuader, 2), "Promedio:", round(promCuader, 2))
````

**Preguntas 5:**
* ¿Cuántas **unidades** vendió cada sucursal por cada categoría?

**Instrucciones:**
* Presente una **tabla cruzada** con totales por fila y por columna.
* Compruebe que el **total general** coincide con la suma de `cantidad` en `ventas.csv`.

````python

datos = pd.merge(ven, pro[["id_producto", "categoria"]], on="id_producto")

tablaCta = pd.crosstab(
    datos["sucursal"],
    datos["categoria"],
    values=datos["cantidad"],
    aggfunc="sum",
    margins=True,
    margins_name="Total"
)

display(tablaCta)

totalVentas = ven["cantidad"].sum()

print("El total de unidades vendidas es:", totalVentas)
print("El total de la tabla es:", tablaCta.loc["Total", "Total"])
````

**Preguntas 6:**
* ¿Cuál fue el **importe total de ventas** de cada sucursal?

> **Nota:** Calcule primero el precio total de cada línea como `cantidad × precio_unitario`.

````python
cont1 = 0
cont2 = 0
cont3 = 0

for ventas, sucursal, precio in zip(ven["cantidad"], ven["sucursal"], ven["precio_unitario"]):
    if sucursal == "Sur":
        cont1 += ventas * precio
    elif sucursal == "Norte":
        cont2 += ventas * precio
    else:
        cont3 += ventas * precio

print("El importe total de Sur es:", round(cont1, 2))
print("El importe total de Norte es:", round(cont2, 2))
print("El importe total de Centro es:", round(cont3, 2))
````

**Preguntas 7:**
* ¿Cuántos **tickets** corresponden a cada rango de unidades compradas?
  * **Pequeñas:** Hasta 7 unidades.
  * **Medianas:** De 8 a 14 unidades.
  * **Grandes:** De 15 o más unidades.

> **Nota:** Sume primero las cantidades de todas las líneas correspondientes a cada ticket.

````python
co1 = 0
co2 = 0
co3 = 0

tickets = ven.groupby("id_ticket")["cantidad"].sum()

for unidades in tickets:
    if unidades <= 7:
        co1 += 1
    elif unidades <= 14:
        co2 += 1
    else:
        co3 += 1

print("Tickets de hasta 7 unidades:", co1)
print("Tickets de 8 a 14 unidades:", co2)
print("Tickets de 15 unidades o más:", co3)
````

**Preguntas 8:**
* ¿Cuántos **tickets** y qué **importe total** se registraron en cada mes?
* ¿En qué mes se alcanzó el **mayor importe**?

````python
import pandas as pd

ven["fecha"] = pd.to_datetime(ven["fecha"])

ven["mes"] = ven["fecha"].dt.month

ven["importe"] = ven["cantidad"] * ven["precio_unitario"]

ticketsMes = ven.groupby("mes")["id_ticket"].nunique()
importeMes = ven.groupby("mes")["importe"].sum()

tablaMes = pd.DataFrame({
    "Tickets": ticketsMes,
    "Importe total": importeMes
})

display(tablaMes)

mesMayor = importeMes.idxmax()
mayorImporte = importeMes.max()

print("El mes con mayor importe de ventas es:", mesMayor)
print("El importe total de ese mes es:", round(mayorImporte, 2))
````

**Preguntas 9:**
* ¿Qué **producto** acumuló más unidades vendidas en cada sucursal?

**Instrucciones:**
* Indique su **código**, **nombre** y **cantidad vendida**.
* En caso de **empate**, incluya todos los productos empatados.

````python
datos = pd.merge(ven, pro[["id_producto", "producto"]], on="id_producto")

ventasProducto = datos.groupby(
    ["sucursal", "id_producto", "producto"]
)["cantidad"].sum()

tablaProductos = ventasProducto.reset_index()

for sucursal in tablaProductos["sucursal"].unique():

    productosSucursal = tablaProductos[tablaProductos["sucursal"] == sucursal]

    mayorVenta = productosSucursal["cantidad"].max()

    mejoresProductos = productosSucursal[productosSucursal["cantidad"] == mayorVenta]

    print("\nSucursal:", sucursal)
    display(mejoresProductos)
````

Por ejemplo, no es lo mismo contar las líneas de venta, contar los tickets distintos o sumar las unidades compradas por cada ticket. Aunque estos procedimientos trabajan con el mismo archivo, cada uno responde a una necesidad diferente.

**3. Aplicación de funciones y métodos de Pandas**

Durante el examen utilicé estructuras como `for`, `if`, `elif`, `else`, listas y contadores, que me permitieron desarrollar procedimientos mediante la lógica que ya conocía. Sin embargo, también identifiqué que algunas operaciones podían realizarse de manera más adecuada utilizando funciones de Pandas.

Entre las herramientas que reforcé se encuentran:

- `groupby()`: permite agrupar registros según una o varias columnas.
- `sum()`: permite sumar los valores de una columna o de un grupo.
- `nunique()`: permite contar los valores distintos sin repetirlos.
- `merge()`: permite relacionar dos DataFrames mediante una columna identificadora.
- `drop_duplicates()`: permite eliminar registros repetidos según una columna.
- `max()` y `min()`: permiten obtener los valores máximos y mínimos.
- `reset_index()`: permite reorganizar los resultados de una agrupación.
- `agg()`: permite calcular varias operaciones sobre grupos de datos.

Aprendí que estas funciones no sustituyen la lógica de programación, sino que permiten aplicarla de una manera más organizada y eficiente cuando se trabaja con DataFrames.
