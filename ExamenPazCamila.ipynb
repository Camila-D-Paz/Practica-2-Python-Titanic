
Tickets por sucursal. ¿Cuántos tickets se registraron en cada sucursal? ¿Cuál registró más?



```python
contTS=0
contTN=0
contTC=0

for surcursal in ven["sucursal"]:
    if surcursal=="Sur":
        contTS+=1
    elif surcursal=="Norte":
        contTN+=1
    else:
        contTC+=1

print("Se vendio en el Sur esta cantidad: ",contTS)
print("Se vendio en el Norte esta cantidad: ",contTN)
print("Se vendio en el Centro esta cantidad: ",contTC)
```

Métodos de pago. ¿Cuántos tickets se pagaron con cada método y cuántos tienen ese dato sin registrar?



```python
contEfec=0
contTarjet=0
contTrans=0
contSinn=0

for metodo in ven["metodo_pago"]:
    if metodo=="Transferencia":
        contTrans+=1
    elif metodo=="Tarjeta":
        contTarjet+=1
    elif metodo=="Efectivo":
        contEfec+=1
    else:
        contSinn+=1

print ("Con el metodo transferencia se tiene: ",contTrans)
print ("Con el metodo Efectivo se tiene: ",contEfec)
print ("Con el metodo Tarjeta se tiene: ",contTarjet)
print ("Sin metodo se tiene: ",contSinn)
```

Precios de lista por categoría. ¿Cuál es el precio de lista mínimo, máximo y promedio de los productos de cada categoría? Use el catálogo de productos.



```python
precioE=[]
precioAr=[]
precioAcce=[]
precioCuader=[]

for categoria,precio in zip(pro["categoria"],pro["precio_lista"]):
    if categoria=="Escritura":
        precioE.append(precio)
    elif categoria=="Archivo":
        precioAr.append(precio)
    elif categoria=="Accesorios":
        precioAcce.append(precio)
    else:
        precioCuader.append(precio)

promE=sum(precioE)/len(precioE)
minE=min(precioE)
maxE=max(precioE)

promAR=sum(precioAr)/len(precioAr)
minAr=min(precioAr)
maxAr=max(precioAr)

promAC=sum(precioAcce)/len(precioAcce)
minAC=min(precioAcce)
maxAC=max(precioAcce)

promC=sum(precioCuader)/len(precioCuader)
minC=min(precioCuader)
maxC=max(precioCuader)

print ("En la categoria Escritura, tiene de maximo: ",maxE,",min: ",minE, " y prmedio: ",promE)
print ("En la categoria Archivo, tiene de maximo: ",maxAr,",min: ",minAr, " y prmedio: ",promAR)
print ("En la categoria Accesorio, tiene de maximo: ",maxAC,",min: ",minAC, " y prmedio: ",promAC)
print ("En la categoria Cuaderno, tiene de maximo: ",maxC,",min: ",minC, " y prmedio: ",promC)
```
